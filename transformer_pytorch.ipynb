```python
import torch
from torch import nn
```

```python
torch.device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using {torch.device} device")
```

```python
class Transformer(nn.Module):
    def __init__(self, vocab_size, embedding_dim, n_heads, n_layers, dropout):
        super().__init__()
        self.vocab_size = vocab_size
        self.embedding_dim = embedding_dim
        self.n_heads = n_heads
        self.n_layers = n_layers
        self.dropout = dropout

        # Definição da camada/layer embedding, transforma a sequencia de entrada em uma sequencia de vetores de
        # dimensão fixa
        self.embedding = nn.Embedding(vocab_size, embedding_dim)

        # Definição do mecanismo de atenção
        self.attention = nn.MultiheadAttention(embedding_dim, n_heads, dropout=dropout)

        # Definição da rede neural feed foward que será usada para gerar a sequencia de saida atraves da entrada
        self.feed_foward = nn.Sequential(
            nn.Linear(embedding_dim, embedding_dim),
            nn.ReLU(),
            nn.Linear(embedding_dim, embedding_dim),
        )

        # Definição da camada de output
        self.out = nn.Linear(embedding_dim, vocab_size)

    def forward(self, x):
        x = self.embedding(x)
        x = self.attention(x)
        x = self.feed_foward(x)
        x = self.out(x)
        return x
```

```python
modelo = Transformer(
    vocab_size=1000,
    embedding_dim=32,
    n_heads=4,
    n_layers=2,
    dropout=0.5,
)
```
