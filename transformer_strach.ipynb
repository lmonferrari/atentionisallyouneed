```python
import numpy as np
```

```python
# dimensão do modelo
dim_model = 64

# tamanho da sequencia do modelo
seq_len = 10

# tamanho do vocabulário
vocab_size = 100
```

```python
def embeding(input, vocab_size, dim_model):

    # cria matriz de embedding onde cada linha representa um token do vocabulário
    # A matriz é inicializada com valores aleatórios normalmente distribuidos
    embed = np.random.randn(vocab_size, dim_model)

    # para cada indice de token no input, seleciona o embedding correspondente da matriz
    # Retorna o array de embedding correspondentes a sequencia de entrada
    return np.array([embed[i] for i in input])
```

```python
def softmax(x):
    e_x = np.exp(x - np.max(x))

    return e_x / e_x.sum(axis=1).reshape(-1, 1)
```

```python
def scaled_dot_product_attention(Q, K, V):
    # calcula  o produto escalar entre Q e a transposta de K
    matmul = np.dot(Q, K.T)

    # obtem a dimensão dos vetores de chave
    depth = K.shape[-1]

    # escala os logits dividindo-os pela raiz quadrada da profundidade
    logits = matmul / np.sqrt(depth)

    # aplica a função softmax para obter os pesos de atenção
    attention_weights = softmax(logits)

    # multiplica os pesos de atenção pelos valores V para obter a saída
    output = np.dot(attention_weights, V)

    # retorna a saída ponderada
    return output
```

```python
def linear_and_softmax(input):
    # Inicializa a matriz de pesos com valores aleatórios normalmente distribuidos
    weights = np.random.randn(dim_model, vocab_size)

    # realiza a operação linear entre entrada e a matriz de pesos
    logits = np.dot(input, weights)

    # aplica o softmax aos logits e retorna
    return softmax(logits)
```

```python
def transform_model(input):
    embedded_input = embeding(input, vocab_size, dim_model)
    attention_output = scaled_dot_product_attention(
        embedded_input, embedded_input, embedded_input
    )
    output_probabilities = linear_and_softmax(attention_output)
    output_indices = np.argmax(output_probabilities, axis=1)

    return output_indices
```

```python
input_sequence = np.random.randint(0, vocab_size, seq_len)
print("Sequencia de entrada: ", input_sequence)
```

```python
output_transformer = transform_model(input_sequence)
print("Saída do modelo: ", output_transformer)
```
